### FILTERING AN BOOLEAN INDEXING

```python
df
```

```python
#simple condition-filter rows matching one condition
df[df['age']>28]
```

```python
#multiple condition-combine with & (and)/|(or)
df[(df['age']>25) &(df['city']== 'NY')]
```

```python
df[(df['age']==35) | (df['city']== 'SF')]
```

```python
#name=bob,or city=ny
df[(df['name']=='Bob') | (df['city']== 'NY')]
```

```python
#.isin()-filter rows where a column matches a set of values
df[df['city'].isin(['NY','LA'])]
```

```python
df.query('age>28 and city=="NY"') #.query
```

```python
#.here()-keep matching values,replace other with NAN
df.where(df['age']>28)
```

```python
#.maskopposite to where(),replacing matching values with Nan
df.mask(df['age']>28)
```

### HANDLING MISSING DATA

```python
df_na=pd.DataFrame({'a':[1,np.nan,3],'b':[np.nan,2,3]})
df_na
```

```python
#.isna()-boolean mask of missing values
df_na.isna()
```

```python
#.notna()-opposite of .isna()
df_na.notna()
```

```python
#.isnull().sum()-count missing values per column
df_na.isnull().sum()
```

```python
#.dropna()-dropping null values/drop rows containing any nan
df_na.dropna()
```

```python
#.dropna(axis=1)-drop column containing any nan
df_na.dropna(axis=1)
```

```python
#.fillna()-fill missing values with a constant
df_na.fillna(0)
```

```python
#.ffill()-forward fill missing values
df_na.ffill()
```

```python
#.bfill()-backward fill missing values
df_na.bfill()
```

### GROUPBY OPERATIONS

```python
df
```

```python
g=df.groupby('city')
```

```python
#.size()-number of rows in each group
g.size()
```

```python
#.group then aggregates a single column
g['age'].mean()
```

```python
#.agg() on a groupby-multiply aggregations  per group
g.agg({'age':['mean','sum']})
```

```python
#.apply() on a groupby-custom  function per group
g.apply(lambda x: x['age'].max() - x['age'].min())
```

```python
add(4,5)
```

```python
iterating over groups with a for loop
```
